### Studi Kasus - Hitung Nilai Akhir dan Nilai Huruf

```python
# import library
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# load dataset sebagai masukan sistem
dataset = pd.read_csv("../dataset/dummy_hitung_nilai_input.csv")
print(dataset)
```

```python
# metadata dari dataset hitung nilai
print(dataset.info())
```

```python
# proses menghitung nilai akhir
nilai_akhir = dataset["formatif"]*0.2 + dataset["uts"]*0.4 + dataset["uas"]*0.4

# menambahkan kolom nilai akhir kedalam dataset
dataset["nilai_akhir"] = nilai_akhir

# menampilkan dataset
print(dataset)
```

```python
# fungsi menghitung nilai huruf
def nilai_huruf(nilai_akhir):
    huruf = []
    for x in nilai_akhir:
        if x >= 80:
            huruf.append("A")
        elif x >= 70:
            huruf.append("B")
        elif x >= 60:
            huruf.append("C")
        elif x >= 50:
            huruf.append("D")
        else:
            huruf.append("E")
    return huruf
```

```python
# proses menghitung nilai huruf
nilai_huruf = nilai_huruf(np.array(dataset["nilai_akhir"]))

# menambahkan kolom nilai huruf kedalam dataset
dataset["nilai_huruf"] = nilai_huruf
```

```python
# Output Program - Pertama
# Dataset miliki dua kolom baru yaitu nilai akhir dan nilai huruf
print(dataset)
```

```python
# Output Program - Kedua
# Visualisasi Data menggunakan barplot untuk melihat sebaran nilai huruf

# membuat figure dengan ukuran 8 x 4
fig, ax = plt.subplots(figsize=(8,4))

# membuat barplot dengan seaborn
sns.barplot(dataset, x="nilai_huruf", y="nilai_akhir")

# add labels
ax.set_title("")
ax.set_xlabel("")
ax.set_ylabel("")

# menampilkan hasilnya
plt.grid(True)
plt.show()
```
